```python
import os
import pandas as pd
data_path='/Users/kimminkyoung/Desktop/hw4/'
forders=os.listdir(data_path)
forders
```

```python
import numpy as np
#하나의 데이터 프레임으로 합치고, 출력
#index번호 제대로 설정됨
df= pd.DataFrame()
print(forders)
for i in forders:
    path=data_path+i
    if path.endswith('.xlsx'):
        add=pd.read_excel(path)
    #print(path)
        data=pd.concat([df,add])
data
```

```python
#일자, 언론사, 기고자, 제목만 따로 데이터프레임으로 추출
data1=data[['일자','언론사','기고자','제목']]
data1
```

```python
#하고 duplicate 데이터를 삭제한 후 데이터프레임을 출력
data2=data1.drop_duplicates()
data2
```

```python
#index
data2=data2.reset_index()
data2
```

```python
data2=data2.drop('index',axis=1)
data2
```

```python
#일자 format 변경
data2['일자']=pd.to_datetime(data2['일자'],format="%Y%m%d")
data2.dtypes
data2
```

```python
from konlpy.tag import Komoran
tagger=Komoran()
```

```python
#복합명사
```

```python
#제목 부분을 konlpy를 이용, 주요 키워드 리스트로 분해하여 데이터프레임에 칼럼으로 추가하라
#기본 명사 추출
token_nouns=[tagger.nouns(d) for d in data2['제목']]
data2['명사']=token_nouns
data2
```

```python
#사용사 사전 적용
#사용사 사전은 bigran 상위 100개 중에서 최소 10개 이상
import nltk

bglist=[]
for sentence in data2["명사"]:
    bglist += nltk.bigrams(sentence)
bglist
```

```python
from collections import Counter
bgCountDict = Counter(bglist)
mostbglist=bgCountDict.most_common(10)
mostbglist
```

```python
import operator
sortedBgCountDict = dict(sorted(bgCountDict.items(),key=operator.itemgetter(1), reverse=True))
sortedBgCountDict
```

```python
with open(data_path+'user_dic.txt') as f:
    for line in f.readlines():
        print(line.replace('\t','<TAB>'))
```

```python
tagger=Komoran(userdic=data_path+'user_dic.txt')
token_nouns=[tagger.nouns(d) for d in data2["제목"]]
data2['복합명사 포함']=token_nouns
data2
```

```python
#불용어
```

```python
#불용어 처리
#불용어는 전체 단어 빈도수 200개 중에서 최소 10개 이상 포함하라 
token_all_one = [t for tlist in data2['복합명사 포함'] for t in tlist]
token_all_one
```

```python
countDict = Counter(token_all_one)
sortedCountList = sorted(countDict.items(),key=operator.itemgetter(1),reverse=True)
sortedCountList
```

```python
stop_words=['그','것','수','그것','이','때문','번','저','때','듯','거','중']
token_final=[]
for doc in data2['복합명사 포함']:
    wordslist=[w for w in doc if w not in stop_words]
    token_final.append(wordslist)
```

```python
data2['불용어 제외']=token_final
data2
```

```python
#csv 파일 및 pickle 파일로 저장할 것
import pickle
with open(data_path+'스마트도시_tokens.dump', 'wb') as f:
    pickle.dump(data2, f)
```

```python
#파일 다시 읽기
```

```python
with open(data_path+"스마트도시_tokens.dump",'rb') as f:
    doc_df=pickle.load(f)
doc_df
```

```python
#불용어 처리까지 끝난 키워드들로 DTM을 만들어서 출력하라. 
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer(analyzer = lambda x:x)
text_count=vectorizer.fit_transform(doc_df['불용어 제외'])
text_count.todense()
```

```python
#DTM의 인덱스가 날짜인 것 1개, 
import pandas as pd
date_df = pd.DataFrame(text_count.todense(), columns = vectorizer.get_feature_names_out())
date_df.index=doc_df["일자"]
date_df
```

```python
date_df=date_df.groupby('일자').sum()
date_df
```

```python
date_df.loc["단어빈도수 합계"]=date_df.sum() #axis=0을 하면 가로로 합을 구함 
date_df
```

```python
# 각 단어의 빈도수 합계를 기준으로 내림차순으로 정렬
date_df_count=date_df.T.sort_values(by="단어빈도수 합계",ascending=False,)
date_df_count
```

```python
#DTM의 인덱스가 언론사인 것
press_df = pd.DataFrame(text_count.todense(), columns = vectorizer.get_feature_names_out())
press_df.index=doc_df["언론사"]
press_df=press_df.head(200)
```

```python
press_df=press_df.groupby('언론사').sum()
press_df
```

```python
press_df.loc["단어빈도수 합계"]=press_df.sum() #axis=0을 하면 가로로 합을 구함 
press_df
```

```python
# 각 단어의 빈도수 합계를 기준으로 내림차순으로 정렬
press_df_count=press_df.T.sort_values(by="단어빈도수 합계",ascending=False)
press_df_count
```

```python
#불용어  처리 끝난 키워드들을 다 합친 후 가장 빈도수가 높은 20개 키워드를 
#내림차순 막대그래프로 그려라.
#전체, 대표 언론사 3개
```

```python
#전체
plt.figure(figsize=(20,6))
plt.bar(press_df_count.index[:20],press_df_count['단어빈도수 합계'][:20])
plt.title("키워드 빈도수")
plt.show()
```
